```python
import random

# Simple grid self-driving with tabular Q-learning (no functions, no classes)

# Environment settings
ROAD_LENGTH = 10
MAX_STEPS = 100

# Q-learning settings
ALPHA = 0.1       # learning rate
GAMMA = 0.95      # discount
EPS = 0.2         # exploration rate
ACTIONS = 4       # 0=right, 1=up, 2=down, 3=left

# Q-table: dictionary state -> [q0, q1, q2, q3]
Q = {}

# Training
for episode in range(200):
    # Reset environment
    x = 0
    y = 0
    done = False
    total_reward = 0.0
    steps = 0

    while not done and steps < MAX_STEPS:
        state = (x, y)

        # Initialize Q for this state if needed
        if state not in Q:
            Q[state] = [0.0, 0.0, 0.0, 0.0]

        # Choose action: epsilon-greedy
        if random.random() < EPS:
            action = random.randint(0, ACTIONS - 1)
        else:
            q_vals = Q[state]
            action = q_vals.index(max(q_vals))

        # Take action
        if action == 0:      # right
            x += 1
        elif action == 1:    # up
            y += 1
        elif action == 2:    # down
            y -= 1
        elif action == 3:    # left
            x -= 1

        # Reward
        reward = -0.1  # small penalty per step

        # Off-road penalty (road is only y = 0)
        if y != 0:
            reward -= 1.0

        # Check done conditions
        done = False

        # Reached goal
        if x >= ROAD_LENGTH - 1 and y == 0:
            reward += 5.0
            done = True

        # Fell off ends
        if x < 0 or x >= ROAD_LENGTH:
            reward -= 2.0
            done = True

        next_state = (x, y)
        if next_state not in Q:
            Q[next_state] = [0.0, 0.0, 0.0, 0.0]

        # Q-learning update
        current_q = Q[state][action]
        if done:
            target = reward
        else:
            next_max_q = max(Q[next_state])
            target = reward + GAMMA * next_max_q

        new_q = current_q + ALPHA * (target - current_q)
        Q[state][action] = new_q

        total_reward += reward
        steps += 1

    if (episode + 1) % 20 == 0:
        print("Episode", episode + 1, "/", 200, " | Steps:", steps, " | Total reward:", round(total_reward, 2))

print("Training complete.")

# Show learned policy for states on the road (y = 0)
action_names = ["right", "up", "down", "left"]

print("\nLearned policy (best action per state on the road):")
for xx in range(ROAD_LENGTH):
    s = (xx, 0)
    if s in Q:
        best_action = Q[s].index(max(Q[s]))
        print("State (x =", xx, ", y = 0): best action =", action_names[best_action])
```
